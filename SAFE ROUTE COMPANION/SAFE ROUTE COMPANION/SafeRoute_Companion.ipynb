```python
class Journey:
    def __init__(self, start_location, destination, expected_arrival):
        self.start_location = start_location
        self.destination = destination
        self.expected_arrival = expected_arrival
        self.status = "Not Started"
        self.safe_check_ins = 0

    def start_journey(self):
        self.status = "In Progress"
        print("Journey started successfully!")

    def confirm_safety(self):
        if self.status == "In Progress":
            self.safe_check_ins += 1
            print("Safety check-in successful!")
            print("You are safe.")
        else:
            print("Please start your journey first.")

    def display_journey(self):
        print("\n--- JOURNEY DETAILS ---")
        print("Start Location:", self.start_location)
        print("Destination:", self.destination)
        print("Expected Arrival:", self.expected_arrival)
        print("Journey Status:", self.status)
        print("Safety Check-ins:", self.safe_check_ins)
```

```python
class Notification:
    def __init__(self, contact_name, contact_number):
        self.contact_name = contact_name
        self.contact_number = contact_number

    def send_alert(self, message):
        print("\n--- EMERGENCY ALERT ---")
        print("Contact:", self.contact_name)
        print("Message:", message)
        print("Status: Simulated notification only")
```

```python
class SafeRouteApp:
    def __init__(self, user, journey, notification):
        self.user = user
        self.journey = journey
        self.notification = notification

    def start_trip(self):
        print("\nWelcome to SafeRoute Companion!")
        print("User:", self.user.name)

        self.journey.start_journey()

    def check_safety(self):
        if self.journey.status == "In Progress":
            self.journey.confirm_safety()
        else:
            print("Please start your journey first.")

    def trigger_emergency_alert(self):
        self.notification.send_alert(
            f"Please check on {self.user.name}. "
            "Their journey may need attention."
        )
```

```python

import sqlite3
from datetime import datetime
from IPython.display import display, HTML

with sqlite3.connect("saferoute.db") as db:
    cursor = db.cursor()

    cursor.execute("""
        SELECT
            j.start_location,
            j.destination,
            c.scheduled_time,
            c.is_completed
        FROM journeys j
        LEFT JOIN check_ins c ON c.journey_id = j.id
        WHERE j.id = (
            SELECT id FROM journeys ORDER BY id DESC LIMIT 1
        )
        ORDER BY c.id DESC
        LIMIT 1
    """)

    reminder_data = cursor.fetchone()

if reminder_data:
    start, destination, scheduled_time, completed = reminder_data

    if completed == 1:
        title = "Safety Check-in Completed"
        message = "You have already confirmed that you are safe."
        color = "#15803D"
        symbol = "✓"
    elif not scheduled_time:
        title = "No Check-in Scheduled"
        message = "Please schedule a safety check-in for this journey."
        color = "#64748B"
        symbol = "!"
    else:
        current_time = datetime.now().strftime("%H:%M")
        if current_time >= scheduled_time:
            title = "Check-in Time Reached"
            message = "Please confirm your safety on the website."
            color = "#B45309"
            symbol = "!"
        else:
            title = "Journey in Progress"
            message = f"Your check-in is scheduled for {scheduled_time}."
            color = "#0F766E"
            symbol = "⏱"

    display(HTML(f"""
    <div style="background:#12343B;color:white;padding:25px;
                border-radius:18px;font-family:Arial,sans-serif;">
        <h2>🔔 Safety Reminder Monitor</h2>
        <p>{start} → {destination}</p>
        <div style="background:{color};padding:15px;border-radius:10px;">
            <span style="font-size:22px;">{symbol}</span>
            <b>{title}</b>
        </div>
        <p>{message}</p>
        <p>Current time: {datetime.now().strftime("%H:%M")}</p>
    </div>
    """))
else:
    display(HTML("<h3>Please create a journey before checking reminders.</h3>"))
```

```python

from flask import Flask, render_template_string, request, redirect, url_for
import sqlite3

app4 = Flask("SafeRouteCompanionV4")

def home4():
    message = request.args.get("message", "")

    with sqlite3.connect("saferoute.db") as db:
        cursor = db.cursor()

        cursor.execute("""
            SELECT u.name, j.id, j.start_location, j.destination,
                   j.expected_arrival, j.status
            FROM journeys j
            JOIN users u ON u.id = j.user_id
            ORDER BY j.id DESC
            LIMIT 1
        """)
        journey = cursor.fetchone()

        cursor.execute("""
            SELECT start_location, destination, expected_arrival, status
            FROM journeys
            ORDER BY id DESC
        """)
        journeys = cursor.fetchall()

    if not journey:
        return "No journey found. Please create a user profile and journey first."

    name, journey_id, start, destination, arrival, status = journey

    return render_template_string("""
    <!DOCTYPE html>
    <html>
    <head>
        <title>SafeRoute Companion</title>
        <meta name="viewport" content="width=device-width, initial-scale=1">
        <style>
            body { margin:0; font-family:Arial,sans-serif;
                   background:#F0F7F7; color:#183B42; }
            header { background:#12343B; color:white; padding:25px 7%; }
            main { max-width:950px; margin:25px auto; padding:0 18px; }
            .card { background:white; padding:24px; border-radius:16px;
                    margin-bottom:20px; }
            h2 { color:#176B70; }
            input { width:100%; box-sizing:border-box; padding:12px;
                    margin:8px 0 16px; border:1px solid #CBD5E1;
                    border-radius:8px; }
            button { background:#0F766E; color:white; padding:13px 18px;
                     border:0; border-radius:9px; cursor:pointer; }
            .safe { background:#15803D; }
            .message { background:#DCFCE7; color:#166534; padding:14px;
                       border-radius:10px; margin-bottom:20px; }
            .table-wrap { overflow-x:auto; }
            table { width:100%; border-collapse:collapse; min-width:450px; }
            th,td { padding:12px; border-bottom:1px solid #E2E8F0;
                    text-align:left; }
        </style>
    </head>
    <body>
        <header>
            <h1>🛡️ SafeRoute Companion</h1>
            <p>Your journey, your safety, your peace of mind.</p>
        </header>
        <main>
            {% if message %}
            <div class="message">{{ message }}</div>
            {% endif %}

            <section class="card">
                <h2>Welcome, {{ name }}!</h2>
                <p><b>From:</b> {{ start }}</p>
                <p><b>Destination:</b> {{ destination }}</p>
                <p><b>Expected arrival:</b> {{ arrival }}</p>
                <p><b>Journey status:</b> {{ status }}</p>
                <form action="/confirm-safety-v4" method="POST">
                    <button class="safe" type="submit">I'm Safe ✓</button>
                </form>
            </section>

            <section class="card">
                <h2>🗺️ Plan a New Journey</h2>
                <form action="/start-journey-v4" method="POST">
                    <label>Starting Location</label>
                    <input name="start_location" required>
                    <label>Destination</label>
                    <input name="destination" required>
                    <label>Expected Arrival</label>
                    <input type="time" name="expected_arrival" required>
                    <label>Scheduled Safety Check-in</label>
                    <input type="time" name="scheduled_checkin" required>
                    <button type="submit">Start Journey</button>
                </form>
            </section>

            <section class="card">
                <h2>📍 Journey History</h2>
                <div class="table-wrap">
                    <table>
                        <tr><th>From</th><th>Destination</th>
                            <th>Arrival</th><th>Status</th></tr>
                        {% for item in journeys %}
                        <tr>
                            <td>{{ item[0] }}</td>
                            <td>{{ item[1] }}</td>
                            <td>{{ item[2] }}</td>
                            <td>{{ item[3] }}</td>
                        </tr>
                        {% endfor %}
                    </table>
                </div>
            </section>
        </main>
    </body>
    </html>
    """, name=name, start=start, destination=destination,
       arrival=arrival, status=status, journeys=journeys, message=message)

def confirm_safety4():
    with sqlite3.connect("saferoute.db") as db:
        cursor = db.cursor()
        cursor.execute("""
            SELECT id FROM check_ins
            WHERE journey_id = (
                SELECT id FROM journeys ORDER BY id DESC LIMIT 1
            )
            ORDER BY id DESC LIMIT 1
        """)
        record = cursor.fetchone()

        if record:
            cursor.execute(
                "UPDATE check_ins SET is_completed = 1 WHERE id = ?",
                (record[0],)
            )
            message = "Safety check-in completed successfully!"
        else:
            message = "No check-in was scheduled for the latest journey."

    return redirect(url_for("home4", message=message))

def start_journey4():
    start = request.form.get("start_location", "").strip()
    destination = request.form.get("destination", "").strip()
    arrival = request.form.get("expected_arrival", "").strip()
    checkin = request.form.get("scheduled_checkin", "").strip()

    if not all([start, destination, arrival, checkin]):
        return "Please complete all journey fields.", 400

    with sqlite3.connect("saferoute.db") as db:
        cursor = db.cursor()
        cursor.execute("SELECT id FROM users ORDER BY id DESC LIMIT 1")
        user = cursor.fetchone()

        if not user:
            return "Please create a user profile first.", 400

        cursor.execute("""
            INSERT INTO journeys
            (user_id, start_location, destination, expected_arrival, status)
            VALUES (?, ?, ?, ?, ?)
        """, (user[0], start, destination, arrival, "In Progress"))

        journey_id = cursor.lastrowid

        cursor.execute("""
            INSERT INTO check_ins (journey_id, scheduled_time, is_completed)
            VALUES (?, ?, 0)
        """, (journey_id, checkin))

    return redirect(url_for("home4", message="Your new journey has been saved!"))

app4.add_url_rule("/", "home4", home4, methods=["GET"])
app4.add_url_rule(
    "/confirm-safety-v4", "confirm_safety4",
    confirm_safety4, methods=["POST"]
)
app4.add_url_rule(
    "/start-journey-v4", "start_journey4",
    start_journey4, methods=["POST"]
)

print("Confirmation-enabled website is ready!")
```

```python

import sqlite3
import pandas as pd
from IPython.display import display, HTML

required_tables = [
    "users",
    "journeys",
    "check_ins",
    "notifications",
    "emergency_contacts"
]

results = []

with sqlite3.connect("saferoute.db") as db:
    cursor = db.cursor()

    cursor.execute("""
        SELECT name
        FROM sqlite_master
        WHERE type = 'table'
    """)

    existing_tables = [row[0] for row in cursor.fetchall()]

    for table in required_tables:
        if table in existing_tables:
            cursor.execute(f"SELECT COUNT(*) FROM {table}")
            count = cursor.fetchone()[0]
            results.append({
                "Database Table": table,
                "Status": "Available",
                "Records": count
            })
        else:
            results.append({
                "Database Table": table,
                "Status": "Missing",
                "Records": 0
            })

display(HTML("""
<div style="background:#12343B;color:white;padding:20px;
            border-radius:14px;font-family:Arial,sans-serif;">
    <h2 style="margin:0;">🔍 SafeRoute Database Health Check</h2>
    <p>Checking the tables used by the project.</p>
</div>
"""))

display(pd.DataFrame(results))
```

```python

import sqlite3
import pandas as pd
from IPython.display import display, HTML

issues = []

with sqlite3.connect("saferoute.db") as db:
    journeys = pd.read_sql_query("""
        SELECT id, user_id, start_location, destination,
               expected_arrival, status
        FROM journeys
    """, db)

    checkins = pd.read_sql_query("""
        SELECT id, journey_id, scheduled_time, is_completed
        FROM check_ins
    """, db)

    contacts = pd.read_sql_query("""
        SELECT id, user_id, contact_name, phone_number
        FROM emergency_contacts
    """, db)

for _, row in journeys.iterrows():
    if pd.isna(row["start_location"]) or not str(row["start_location"]).strip():
        issues.append(f"Journey {row['id']}: Missing starting location")
    if pd.isna(row["destination"]) or not str(row["destination"]).strip():
        issues.append(f"Journey {row['id']}: Missing destination")
    if pd.isna(row["expected_arrival"]) or not str(row["expected_arrival"]).strip():
        issues.append(f"Journey {row['id']}: Missing arrival time")

for _, row in checkins.iterrows():
    if row["is_completed"] not in [0, 1]:
        issues.append(f"Check-in {row['id']}: Invalid completion value")
    if row["journey_id"] not in journeys["id"].values:
        issues.append(f"Check-in {row['id']}: Journey not found")

for _, row in contacts.iterrows():
    if pd.isna(row["contact_name"]) or not str(row["contact_name"]).strip():
        issues.append(f"Contact {row['id']}: Missing contact name")
    if pd.isna(row["phone_number"]) or not str(row["phone_number"]).strip():
        issues.append(f"Contact {row['id']}: Missing phone number")

display(HTML("""
<div style="background:#12343B;color:white;padding:20px;
            border-radius:14px;font-family:Arial,sans-serif;">
    <h2 style="margin:0;">🧪 Final Consistency Test</h2>
    <p>Checking stored records for common data problems.</p>
</div>
"""))

print("Journeys checked:", len(journeys))
print("Check-ins checked:", len(checkins))
print("Emergency contacts checked:", len(contacts))

if issues:
    print("\nIssues found:")
    for issue in issues:
        print("-", issue)
else:
    print("\nPASS: No issues found by these checks!")
```

```python

import sqlite3
import pandas as pd
from IPython.display import display, HTML

with sqlite3.connect("saferoute.db") as db:
    results = pd.read_sql_query("""
        SELECT
            j.id AS journey_id,
            j.start_location,
            j.destination,
            j.status AS journey_status,
            COUNT(c.id) AS total_checkins,
            SUM(CASE WHEN c.is_completed = 1 THEN 1 ELSE 0 END)
                AS completed_checkins
        FROM journeys j
        LEFT JOIN check_ins c ON j.id = c.journey_id
        GROUP BY j.id
        ORDER BY j.id DESC
    """, db)

display(HTML("""
<div style="background:#12343B;color:white;padding:18px;
            border-radius:12px;font-family:Arial,sans-serif;">
    <h2 style="margin:0;">Journey Status Review</h2>
    <p style="margin-bottom:0;">Checking journey and check-in records</p>
</div>
"""))

display(results)

print("\nReview complete. No database records were changed.")
```

```python

import sqlite3
import pandas as pd
from IPython.display import display, HTML

with sqlite3.connect("saferoute.db") as db:
    summary = pd.read_sql_query("""
        SELECT
            j.id AS journey_id,
            j.start_location,
            j.destination,
            j.status AS journey_status,
            COUNT(c.id) AS total_checkins,
            COALESCE(SUM(c.is_completed), 0) AS completed_checkins
        FROM journeys j
        LEFT JOIN check_ins c ON j.id = c.journey_id
        GROUP BY j.id
        ORDER BY j.id DESC
    """, db)

display(HTML("""
<div style="background:#12343B;color:white;padding:20px;
            border-radius:14px;font-family:Arial,sans-serif;">
    <h2 style="margin:0;">🛡️ Journey Safety Summary</h2>
    <p style="margin-bottom:0;">A clear view of journey and check-in progress</p>
</div>
"""))

if summary.empty:
    print("No journeys have been recorded yet.")
else:
    for _, row in summary.iterrows():
        display(HTML(f"""
        <div style="background:#F0F7F7;padding:16px;
                    margin-top:12px;border-radius:12px;
                    border-left:5px solid #0F766E;
                    font-family:Arial,sans-serif;color:#183B42;">
            <h3 style="margin-top:0;">
                {row['start_location']} → {row['destination']}
            </h3>
            <p><b>Journey status:</b> {row['journey_status']}</p>
            <p><b>Completed check-ins:</b>
                {int(row['completed_checkins'])}
                / {int(row['total_checkins'])}</p>
        </div>
        """))
```
